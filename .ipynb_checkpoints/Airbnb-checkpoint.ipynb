```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("compressed_data.csv")
```

```python
df.columns
```

# Checking missing values

```python
print(df.isnull().sum())
```

```python
df.info()
```

# Handling Missing Values

```python
df["last review"] = pd.to_datetime(df["last review"], errors = 'coerce')
```

```python
df.info()
```

```python
df.fillna({'reviews per month': 0, 'last review' : df['last review'].min()},inplace = True)
```

```python
df.dropna(subset = {'NAME','host name'}, inplace = True)
```

```python
print(df.isnull().sum())
```

```python
df = df.drop(columns =['license','house_rules'],errors='ignore')
```

```python
df["price"]=df["price"].replace('[\$,]','', regex = True).astype(float)
df["service fee"] = df["service fee"].replace('[\$,]','',regex =True).astype(float)
```

```python
df.head()
```

# Removing $ and changing datatype to Float 

```python
df['price'] = df['price'].replace(['\$'],'', regex = True).astype(float)
df['service fee'] =df['service fee'].replace(['\$'],'', regex = True).astype(float)
```

```python
df.head()
```

# Removing duplicates

```python
df.drop_duplicates(inplace =True)
```

```python
df.info()
```

# Discriptive Statistics

```python
df.describe()
```

# Visualization

# What is the distribution of listing prices?

```python
plt.figure(figsize =(10,6))
sns.histplot(df['price'],bins =50, kde = True, color = 'red')
plt.title("Distribution of Listing Price")
plt.xlabel("price $")
plt.ylabel("Frequency")
```

### This histogram shows a fairly even distirbution of listing price across different price ranges, indicating no particular concentration of listing in any specific price range. The KDE line helps visualize this even spread more clearly, confirming that dataset contains listings with a wide varity of prices.

# How are diffrent room types Distributed

```python
df.columns
```

```python
df["room type"]
```

```python
plt.figure(figsize =(8,5))
sns.countplot(x= "room type", data =df, color ="green")
plt.title("Room Type distribution")
plt.xlabel("Room Type")
plt.ylabel("Count")
plt.show()
```

# How are listing distributed accross different neighorhood?

```python
plt.figure(figsize =(12,8))
sns.countplot(x ="neighbourhood group" , data = df, color = "Brown" , order=df["neighbourhood group"].value_counts().index)
plt.title("Number of listing By Neighbourhood group")
plt.xlabel("Neighbourhood group")
plt.ylabel("Count")
plt.show()
```

# what is the relationship between price and room type

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    x="room type",
    y="price",
    hue="room type",
    data=df,
    palette="Set1"
)

plt.title("Price VS Room Type")
plt.xlabel("Room Type")
plt.ylabel("Price")
plt.legend(title="Room Type")

plt.show()
```

# How has the number of reviews change over with time

```python
df["last review"] = pd.to_datetime(df["last review"])
reviews_over_time = df.groupby(df["last review"].df.to_period("M")).size()

plt.figure(figsize=(10, 6))

reviews_over_time.plot(
   kind ="line",color="red"
)

plt.title("Number of Reviews Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Reviews")


plt.show()
```
